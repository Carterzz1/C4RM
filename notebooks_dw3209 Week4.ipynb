```python
def getBondPrice(y, face, couponRate, m, ppy=1):
    coupon = face * couponRate / ppy
    periods = int(m * ppy)
    bondPrice = 0

    for p in range(1, periods + 1):
        bondPrice += coupon / (1 + y / ppy) ** p

    bondPrice += face / (1 + y / ppy) ** periods
    return bondPrice

```

```python
def getBondDuration(y, face, couponRate, m, ppy=1):
    coupon = face * couponRate / ppy
    periods = int(m * ppy)
    bondPrice = 0
    weightedPV = 0

    for p in range(1, periods + 1):
        cashFlow = coupon

        if p == periods:
            cashFlow += face

        pv = cashFlow / (1 + y / ppy) ** p
        bondPrice += pv
        weightedPV += (p / ppy) * pv

    bondDuration = weightedPV / bondPrice
    return bondDuration

```

```python
def getBondPrice_E(face, couponRate, m, yc):
    coupon = face * couponRate
    bondPrice = 0

    for t, y in enumerate(yc, start=1):
        cashFlow = coupon

        if t == m:
            cashFlow += face

        bondPrice += cashFlow / (1 + y) ** t

    return bondPrice

```

```python
def getBondPrice_Z(face, couponRate, times, yc):
    coupon = face * couponRate
    bondPrice = 0

    for t, y in zip(times, yc):
        bondPrice += coupon / (1 + y) ** t

    bondPrice += face / (1 + yc[-1]) ** times[-1]
    return bondPrice
```

```python
def FizzBuzz(start, finish):
    outlist = []

    for i in range(start, finish + 1):
        if i % 15 == 0:
            outlist.append('fizzbuzz')
        elif i % 3 == 0:
            outlist.append('fizz')
        elif i % 5 == 0:
            outlist.append('buzz')
        else:
            outlist.append(i)

    return outlist
```
